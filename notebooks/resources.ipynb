# AiiDAlab Code Setup

This page provides an interface for creating and managing computer connections and AiiDA code instances for running DL_POLY calculations.

# Create New Code

Use the widget below to register a new computer connection and/or a new DL_POLY code instance. Pre-configured recipes can be applied via **Quick setup**, or a **Manual setup** can be performed for unlisted computers/codes.

```python
from aiidalab_widgets_base.loaders import LoadingWidget
from ipywidgets import VBox

new_codes_widget_container = VBox(
    children=[
        LoadingWidget("Loading resource setup widget"),
    ],
)
display(new_codes_widget_container)
```

# Available codes

Below are the available codes in the AiiDA database. If codes are added or deleted, click **Refresh codes** to update the list.

```python
available_codes_widget_container = VBox(
    children=[
        LoadingWidget("Loading available codes"),
    ],
)
display(available_codes_widget_container)
```

```python
from alc_aiidalab_widgets.widgets import CodeSetupWidget

new_codes_widget_container.children = [CodeSetupWidget()]
```

```python
from home.code_setup import create_paginated_table, fetch_code_data
from ipywidgets import Button, Layout, Output

output = Output()


def render():
    """Fetch the available codes and render them as a paginated table."""
    data = fetch_code_data()
    paginated_table = create_paginated_table(data)
    output.clear_output()
    with output:
        display(paginated_table)


refresh_button = Button(
    description="Refresh codes",
    button_style="primary",
    icon="refresh",
    tooltip="Refresh the list of codes",
    layout=Layout(width="fit-content"),
)
refresh_button.on_click(lambda _: render())

available_codes_widget_container.children = [
    refresh_button,
    output,
]

render()
```
